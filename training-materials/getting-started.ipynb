## 1. Check that R is working

Run the code below. You should see a short message confirming that R is running in your DNAnexus JupyterLab environment.

```r
message("Hello from DNAnexus!")
```

## 2. Read the training data

The synthetic training data are stored in the persistent DNAnexus project, inside the `mcps-dnanexus-onboarding/data/` folder.

The path `/mnt/project/` is how R accesses files stored in the current DNAnexus project.

```r
dat <- read.csv(
  "/mnt/project/mcps-dnanexus-onboarding/data/example_data.csv"
)

head(dat)
```

## 3. Inspect the data

Run the following cells to see how many observations and variables are in the dataset and to obtain a basic summary.

```r
dim(dat)
```

```r
summary(dat)
```

## 4. Create a simple summary table

Here we use base R to calculate the mean age by sex. No additional R packages are required.

```r
aggregate(age ~ sex, data = dat, FUN = mean)
```

## 5. Create a figure

This simple plot shows the relationship between age and systolic blood pressure.

```r
plot(
  dat$age,
  dat$systolic_bp,
  xlab = "Age (years)",
  ylab = "Systolic blood pressure",
  main = "Synthetic training data"
)
```
